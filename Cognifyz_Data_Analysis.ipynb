LEVEL 1 TASK 1

```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset.csv")

# Show first 5 rows
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
# Count top cuisines
top_cuisines = df['Cuisines'].value_counts()

# Show top 5
top_cuisines.head()
```

```python
# Percentage of cuisines
top_cuisines_percentage = (top_cuisines / len(df)) * 100

top_cuisines_percentage.head()
```

```python
import matplotlib.pyplot as plt

top_cuisines.head(5).plot(kind='bar')
plt.title("Top 5 Cuisines")
plt.show()
```

TASK 2

```python
city_analysis = df.groupby('City').agg({
    'Restaurant ID': 'count',
    'Aggregate rating': 'mean'
})

city_analysis.head()
```

```python
city_analysis.columns = ['Restaurant Count', 'Average Rating']

city_analysis.head()
```

```python
city_analysis.sort_values(by='Restaurant Count', ascending=False).head()
```

```python
city_analysis.sort_values(by='Average Rating', ascending=False).head()
```

```python
city_analysis.sort_values(by='Restaurant Count', ascending=False).head(5)['Restaurant Count'].plot(kind='bar')

plt.title("Top Cities by Number of Restaurants")
plt.show()
```

```python
price_range_count = df['Price range'].value_counts()

price_range_count
```

```python
price_range_count = price_range_count.sort_index()

price_range_count
```

```python
delivery_analysis = df.groupby('Has Online delivery').agg({
    'Restaurant ID': 'count',
    'Aggregate rating': 'mean'
})

delivery_analysis
```

TASK 3

```python
import matplotlib.pyplot as plt

plt.scatter(df['Longitude'], df['Latitude'])

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['Longitude'], df['Latitude'], alpha=0.5)

plt.title("Geographic Distribution of Restaurants")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['Longitude'], df['Latitude'], c=df['Aggregate rating'], cmap='viridis')

plt.colorbar(label='Rating')

plt.title("Restaurant Locations by Rating")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.show()
```
